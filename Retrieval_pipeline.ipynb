```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("/content/Movies_Knowledge_Base_5Pages.pdf")

documents = loader.load()

print("Number of Pages:", len(documents))
```

```python
for i, doc in enumerate(documents):
    print(f"\nPAGE {i+1}")
    print(doc.page_content[:500])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)

print("Total Chunks:", len(chunks))
```

```python
for i in range(3):
    print(f"\nCHUNK {i+1}")
    print(chunks[i].page_content)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import FAISS

vector_db = FAISS.from_documents(
    chunks,
    embeddings
)

print("FAISS Database Created Successfully")
```

```python
query = "What are the major movie genres?"

results = vector_db.similarity_search(
    query,
    k=3
)

for i, doc in enumerate(results):
    print(f"\nRESULT {i+1}")
    print("="*60)
    print(doc.page_content)
```

```python
query = "How has cinema evolved over time?"

results = vector_db.similarity_search(
    query,
    k=3
)

for i, doc in enumerate(results):
    print(f"\nRESULT {i+1}")
    print("="*60)
    print(doc.page_content)
```

```python
query = "What is the future of movies?"

results = vector_db.similarity_search(
    query,
    k=3
)

for i, doc in enumerate(results):
    print(f"\nRESULT {i+1}")
    print("="*60)
    print(doc.page_content)
```

```python
query = "history of cinema"

results = vector_db.similarity_search(query, k=3)

for i, doc in enumerate(results):

    print(f"\nRESULT {i+1}")
    print("="*50)

    print("PAGE:", doc.metadata['page'] + 1)

    print(doc.page_content)
```

```python
vector_db.save_local("movie_faiss_db")

print("Database Saved")
```
